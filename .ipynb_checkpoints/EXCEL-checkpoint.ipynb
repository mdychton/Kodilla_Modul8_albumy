```python
writer = pd.ExcelWriter('add_chart.xlsx', engine='xlsxwriter')
df.to_excel(writer, sheet_name='my_chart')
```

```python
workbook = writer.book
worksheet = writer.sheets['my_chart']
```

```python
chart = workbook.add_chart({'type':'line'})
```

```python
def grab_series(df, sheet_name, colname, startcol=0, startrow=0):

    col_index = df.columns.tolist().index(colname)
    col_letter = chr(ord('@')+(col_index+2+startcol))
    first_row = startrow + 2
    last_row = startrow + 1 + len(df)
    return f"='{sheet_name}'!{col_letter}{first_row}:{col_letter}{last_row}"
```

```python
chart.add_series({'values':grab_series(df,'my_chart','B')})
```

```python
chart.set_x_axis({
    'name': 'x^2',
    'name_font': {'size': 14, 'bold': True},
    'num_font':  {'italic': True },
})
```

```python
chart.set_legend({'none': True})
```

```python
worksheet.insert_chart('F2', chart)
```

```python
workbook.close()
```

# ZADANIE


```python
url = "https://www.officialcharts.com/chart-news/the-best-selling-albums-of-all-time-on-the-official-uk-chart__15551/"

tables = pd.read_html(
    url,
    header=0,
    storage_options={
        "User-Agent": "Mozilla/5.0"
    }
)
```

```python
len(tables)
```

```python
df = tables[0]
```

```python
df.columns = df.columns.str.strip() #czyszczenie nazw- usuniecie spacji
```

zmiana nazw wybranych kolumn

```python
df = df.rename(columns={
    'TITLE' : 'TYTUŁ',
    'ARTIST' : 'ARTYSTA',
    'YEAR' : 'ROK',
    'HIGH POSN' : 'MAX POZ'
})
```

```python
df.head() # z defaultu zawsze 5 wierszy 
```

```python
df['ARTYSTA'].nunique() #liczba unikalnych artystow
```

```python
df['ARTYSTA'].value_counts().head() #ile razy pojawia sie dany artysta
```

```python
df.columns = df.columns.str.capitalize()  #headry z duzych liter a reszta malymi
```

```python
df
```

```python
df = df.drop(columns=['Max poz'])  #usuniecie kolumny
```

```python
df
```

```python
df['Rok'].value_counts().head(1)   #w ktorym roku bylo najwiecej albumow
```

```python
df[(df['Rok'] >= 1960) & (df['Rok'] <= 1990)].shape[0]  # ilosc albumow miedzy 1960 a 1990
```

```python
df['Rok'].max()  #najmlodszy album
```

```python
earliest_albums = df.loc[df.groupby('Artysta')['Rok'].idxmin()]  #lista najwczesniej wydanych albumow kazdego artysty
```

```python
earliest_albums_sorted = earliest_albums.sort_values('Rok')
```

```python
earliest_albums = earliest_albums.sort_values('Rok')
```

```python
earliest_albums
```

```python
earliest_albums.to_csv('earliest_albums.csv', index=False) #zapis to pliku CSV (bez indexu)
```
